```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import zipfile
from google.colab import files

# ============================================================
# 1. DATA LOADING AND PREPARATION
# ============================================================

file_name = 'final_summary.csv'

if not os.path.exists(file_name):
    print(f"ERROR: File {file_name} not found. Please upload it using the panel on the left.")
else:
    df = pd.read_csv(file_name)

    def parse_scenario(name):
        architecture = 'Dual Vault' if 'dual' in name else 'Single Vault'
        variant = 'Bez Arbitrażu' if 'no_arb' in name else 'Z Arbitrażem'
        return pd.Series([architecture, variant])

    df[['Architektura', 'Wariant']] = df['scenario'].apply(parse_scenario)
    df['N'] = df['user_count']
    df['Scenariusz'] = df['Architektura'] + ' - ' + df['Wariant']

    sns.set_theme(style="whitegrid")
    plt.rcParams.update({'font.size': 12})

    export_dir = 'wyniki_analizy'
    if not os.path.exists(export_dir):
        os.makedirs(export_dir)

    # ============================================================
    # 2. PLOT GENERATION
    # ============================================================

    # WYKRES 1: Średni spread - wszystkie scenariusze
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df, x='N', y='avg_spread_pct_mean', hue='Scenariusz')
    plt.title('Średni spread cenowy – porównanie wszystkich scenariuszy')
    plt.ylabel('Spread (%)')
    plt.xlabel('Liczba konkurujących agentów (N)')
    plt.legend(title='Scenariusz', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig(f'{export_dir}/wykres_spread.png', dpi=300, bbox_inches='tight')
    plt.close()

    # WYKRES 2: Zysk LP na 1M wolumenu - wszystkie scenariusze
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df, x='N', y='lp_profit_per_1m_vol_mean', hue='Scenariusz')
    plt.title('Zysk dostawców płynności (LP) na 1M wolumenu – porównanie wszystkich scenariuszy')
    plt.ylabel('Zysk (jednostki / 1M wolumenu)')
    plt.xlabel('Liczba konkurujących agentów (N)')
    plt.legend(title='Scenariusz', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig(f'{export_dir}/wykres_zysk_lp.png', dpi=300, bbox_inches='tight')
    plt.close()

    # WYKRES 3: User PnL - wszystkie scenariusze
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df, x='N', y='user_pnl_mean', hue='Scenariusz')
    plt.title('Wynik finansowy użytkowników (User PnL) – porównanie wszystkich scenariuszy')
    plt.ylabel('PnL (im wyżej, tym lepsza ochrona kapitału)')
    plt.xlabel('Liczba konkurujących agentów (N)')
    plt.legend(title='Scenariusz', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig(f'{export_dir}/wykres_user_pnl.png', dpi=300, bbox_inches='tight')
    plt.close()

    # WYKRES 4: Koszt Compute Units - wszystkie scenariusze
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df, x='N', y='avg_cu_per_swap_mean', hue='Scenariusz')
    plt.title('Średni koszt obliczeniowy (Compute Units na swap) – porównanie wszystkich scenariuszy')
    plt.ylabel('Compute Units (CU)')
    plt.xlabel('Liczba konkurujących agentów (N)')
    plt.legend(title='Scenariusz', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig(f'{export_dir}/wykres_koszt_cu.png', dpi=300, bbox_inches='tight')
    plt.close()

    # WYKRES 5: Impermanent Loss - wszystkie scenariusze
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df, x='N', y='impermanent_loss_pct_mean', hue='Scenariusz')
    plt.title('Impermanent Loss (IL) – porównanie wszystkich scenariuszy')
    plt.ylabel('IL (%) – im niższy, tym lepiej')
    plt.xlabel('Liczba konkurujących agentów (N)')
    plt.axhline(y=0, color='red', linestyle='--', linewidth=0.8, alpha=0.5)
    plt.legend(title='Scenariusz', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig(f'{export_dir}/wykres_impermanent_loss.png', dpi=300, bbox_inches='tight')
    plt.close()

    print("Plots have been generated successfully.")

    # ============================================================
    # 3. EXPORT TO ZIP
    # ============================================================

    zip_name = 'wykresy_praca_inz.zip'
    with zipfile.ZipFile(zip_name, 'w') as zipf:
        for root, dirs, files_in_dir in os.walk(export_dir):
            for file in files_in_dir:
                zipf.write(os.path.join(root, file), file)

    print(f"Preparing download: {zip_name}...")
    files.download(zip_name)
```
